# Idiosyncratic Volatility - Public Sample Workflow

Read synthetic returns and factor returns, then estimate toy residual volatility.

This notebook uses only synthetic public sample data from `../input_sample`.


```python
from pathlib import Path
import pandas as pd

base = Path('..')
returns = pd.read_csv(base / 'input_sample' / 'sample_daily_returns.csv')
factors = pd.read_csv(base / 'input_sample' / 'sample_factor_returns.csv')

df = returns.merge(factors, on='date', how='left')
df['toy_residual'] = df['return'] - df['mkt']
ivol = df.groupby('ticker')['toy_residual'].std().rename('toy_ivol').reset_index()
ivol
```
