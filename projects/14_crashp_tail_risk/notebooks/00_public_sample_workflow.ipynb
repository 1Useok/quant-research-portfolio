# CRASHP Tail Risk - Public Sample Workflow

Read synthetic quarterly features, then create a simple tail-risk ranking.

This notebook uses only synthetic public sample data from `../input_sample`.


```python
from pathlib import Path
import pandas as pd

base = Path('..')
panel = pd.read_csv(base / 'input_sample' / 'sample_quarterly_panel.csv')

panel['toy_tail_risk_score'] = 0.4 * panel['tvol'] - 0.3 * panel['tskew'] - 0.2 * panel['rm5']
panel.sort_values('toy_tail_risk_score', ascending=False)[['ticker', 'toy_tail_risk_score', 'crash_label', 'jump_label']]
```
