# ASVI Dividend Attention - Public Sample Workflow

Read synthetic search volume and dividend panel samples, then create a toy attention signal.

This notebook uses only synthetic public sample data from `../input_sample`.


```python
from pathlib import Path
import pandas as pd

base = Path('..')
search = pd.read_csv(base / 'input_sample' / 'sample_search_volume.csv')
panel = pd.read_csv(base / 'input_sample' / 'sample_dividend_panel.csv')

search['asvi_toy'] = search['search_index'] - search['search_index'].rolling(3, min_periods=1).mean()
latest_attention = search['asvi_toy'].iloc[-1]
panel['toy_attention_adjusted_yield'] = panel['dividend_yield'] * (1 + latest_attention / 100)
panel[['ticker', 'dividend_yield', 'toy_attention_adjusted_yield', 'return_1m']]
```
