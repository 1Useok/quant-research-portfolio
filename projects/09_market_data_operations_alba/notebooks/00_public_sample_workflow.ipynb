# Market Data Operations - Public Sample Workflow

Read synthetic universe and minute bars, then run simple quality checks.

This notebook uses only synthetic public sample data from `../input_sample`.


```python
from pathlib import Path
import pandas as pd

base = Path('..')
universe = pd.read_csv(base / 'input_sample' / 'sample_universe.csv')
bars = pd.read_csv(base / 'input_sample' / 'sample_minute_bar.csv')

checks = {
    'universe_rows': len(universe),
    'minute_bar_rows': len(bars),
    'volume_nonnegative': bool((bars['volume'] >= 0).all()),
    'ohlc_valid': bool(((bars['high'] >= bars[['open','close','low']].max(axis=1)) & (bars['low'] <= bars[['open','close','high']].min(axis=1))).all()),
}
checks
```
