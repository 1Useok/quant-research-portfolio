# Quant Subscription Platform - Public Sample Workflow

Read synthetic strategy scores and prices, then form a toy long portfolio.

This notebook uses only synthetic public sample data from `../input_sample`.


```python
from pathlib import Path
import pandas as pd

base = Path('..')
scores = pd.read_csv(base / 'input_sample' / 'sample_strategy_scores.csv')
prices = pd.read_csv(base / 'input_sample' / 'sample_prices.csv')

first = scores[scores['date'] == scores['date'].min()]
selected = first.sort_values('strategy_score', ascending=False).head(3)['ticker']
price_panel = prices[prices['ticker'].isin(selected)]
summary = price_panel.groupby('ticker')['close'].agg(['first', 'last']).reset_index()
summary['toy_return'] = summary['last'] / summary['first'] - 1
summary
```
