# Industry Shock Wave - Public Sample Workflow

Read synthetic input-output and firm panel samples, then combine toy exposure and risk scores.

This notebook uses only synthetic public sample data from `../input_sample`.


```python
from pathlib import Path
import pandas as pd

base = Path('..')
io = pd.read_csv(base / 'input_sample' / 'sample_io_matrix.csv')
firms = pd.read_csv(base / 'input_sample' / 'sample_firm_panel.csv')

industry_exposure = io.groupby('buyer_industry')['input_share'].sum().rename('direct_input_exposure')
result = firms.join(industry_exposure, on='industry')
result['toy_risk_score'] = 100 * (0.6 * result['financial_score'] + 0.4 * result['market_signal'])
result[['firm_id', 'industry', 'direct_input_exposure', 'toy_risk_score']]
```
