# PER Anomaly - Public Sample Workflow

Read synthetic firm characteristics and PER panels, then compute valuation mispricing.

This notebook uses only synthetic public sample data from `../input_sample`.


```python
from pathlib import Path
import pandas as pd

base = Path('..')
chars = pd.read_csv(base / 'input_sample' / 'sample_firm_characteristics.csv')
per = pd.read_csv(base / 'input_sample' / 'sample_per_panel.csv')

df = per.merge(chars, on=['date', 'ticker'], how='left')
df['per_gap'] = df['predicted_per'] - df['actual_per']
df['signal'] = df['per_gap'].rank(ascending=False)
df[['ticker', 'actual_per', 'predicted_per', 'per_gap', 'signal', 'return_next_m']]
```
